```python
from pyspark.sql.types import StructType, StructField, IntegerType, StringType

myschema = StructType(
    [
        StructField("id", IntegerType(), False),
        StructField("name", StringType(), True),
        StructField("age", IntegerType(), True)
    ]
)
```

```python
data = [
    (1, "Dipankar", 27),
    (2, None, 26),
    (3, "Tuhin", None)
]

df = spark.createDataFrame(data, schema=myschema)
df.show()
```

```python
import pyspark
import pkgutil

print("PySpark version:", pyspark.__version__)
print("\nAvailable top-level modules:")

for module in pkgutil.iter_modules(pyspark.__path__):
    print(module.name)
```

```python
import pkgutil
import pyspark

def list_modules(package, prefix="pyspark"):
    print(prefix)

    if hasattr(package, "__path__"):
        for item in sorted(
            pkgutil.iter_modules(package.__path__),
            key=lambda x: x.name
        ):
            full_name = f"{prefix}.{item.name}"
            print(full_name)

            if item.ispkg:
                try:
                    subpackage = __import__(
                        full_name,
                        fromlist=["*"]
                    )
                    list_modules(subpackage, full_name)
                except Exception:
                    # Some modules need optional dependencies
                    # or special runtime configuration.
                    pass

list_modules(pyspark)
```
